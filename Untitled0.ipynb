```python
import importlib
import subprocess
import sys

REQ_FILE = "/content/paper/baseline/SimpleMem/requirements.txt"
MIN_PACKAGES = [
    "openai",
    "numpy",
    "pandas",
    "pyarrow",
    "lancedb",
    "dateparser",
    "tqdm",
    "nltk",
    "rouge-score",
    "bert-score",
    "sentence-transformers",
]

def pip_install(args):
    cmd = [sys.executable, "-m", "pip", "install"] + args
    print("+", " ".join(cmd))
    subprocess.check_call(cmd)

pip_install(["--upgrade", "pip", "setuptools", "wheel"])

try:
    pip_install(["-r", REQ_FILE])
except subprocess.CalledProcessError:
    print("Full requirements install failed; falling back to experiment-critical packages.")
    pip_install(MIN_PACKAGES)

for module_name in [
    "nltk",
    "rouge_score",
    "bert_score",
    "sentence_transformers",
    "lancedb",
    "dateparser",
]:
    importlib.import_module(module_name)

print("Python dependencies ready.")
```

```python
import os
import subprocess
import sys
from pathlib import Path

import nltk

NLTK_DIR = Path("/content/nltk_data")
NLTK_DIR.mkdir(parents=True, exist_ok=True)
os.environ["NLTK_DATA"] = str(NLTK_DIR)

if str(NLTK_DIR) not in nltk.data.path:
    nltk.data.path.insert(0, str(NLTK_DIR))

required_resources = {
    "punkt": "tokenizers/punkt",
    "punkt_tab": "tokenizers/punkt_tab/english",
    "wordnet": "corpora/wordnet",
    "omw-1.4": "corpora/omw-1.4",
}

for pkg, resource in required_resources.items():
    try:
        nltk.data.find(resource)
        print(f"ready: {pkg}")
    except LookupError:
        print(f"downloading: {pkg}")
        nltk.download(pkg, download_dir=str(NLTK_DIR), quiet=False, raise_on_error=True)
        nltk.data.find(resource)

child_code = r'''
import os
import nltk
from nltk.tokenize import word_tokenize
from nltk.translate.meteor_score import meteor_score

print("child NLTK_DATA =", os.environ.get("NLTK_DATA"))
nltk.data.find("tokenizers/punkt_tab/english")
nltk.data.find("corpora/wordnet")
print("child tokenization =", word_tokenize("SimpleMem is ready."))
print("child meteor =", meteor_score([["simplemem", "is", "ready"]], ["simplemem", "is", "ready"]))
'''

subprocess.check_call([sys.executable, "-c", child_code], env=os.environ.copy())
print("NLTK resources are ready for both notebook and subprocesses.")
```
